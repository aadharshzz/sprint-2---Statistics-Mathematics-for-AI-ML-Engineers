## 1. Introduction to Probability

**Probability** is a numerical measure of how likely an event is to occur.

The probability of an event ranges from **0 to 1**:
- 0 → impossible
- 1 → certain
- Between 0 and 1 → different levels of likelihood

For equally likely outcomes:

P(A) = Number of favorable outcomes / Total number of possible outcomes

### Why is it used?
Probability helps us handle uncertainty and estimate how likely different outcomes are.

### When is it used?
It is used in risk analysis, weather forecasting, finance, insurance, quality control, statistics, and AI/ML.

### Real-world analogy
If a bag contains 3 red balls and 2 blue balls, the probability of randomly selecting a red ball is 3/5.

### Program:

```python
# Basic probability example

red_balls = 3
blue_balls = 2

total_balls = red_balls + blue_balls
probability_red = red_balls / total_balls

print("Total balls:", total_balls)
print("Probability of selecting a red ball:", probability_red)
```

### Interpretation
The probability is **0.6**, meaning there is a 60% probability of selecting a red ball when one ball is selected randomly.

## 2. Events

An **event** is a specific outcome or a set of outcomes that we are interested in from a random experiment.

Examples:
- Getting a head when tossing a coin
- Getting an even number when rolling a die
- Selecting a red card

### Why is it used?
Events allow us to describe and calculate the probability of outcomes that matter to us.

### When is it used?
Whenever we perform a probability experiment and want to study a particular outcome or group of outcomes.

### Real-world analogy
If you roll a die, "getting an even number" is an event containing the outcomes 2, 4, and 6.

### Program:

```python
# Event example: rolling an even number

outcomes = [1, 2, 3, 4, 5, 6]
even_event = [x for x in outcomes if x % 2 == 0]

probability_even = len(even_event) / len(outcomes)

print("Sample outcomes:", outcomes)
print("Even-number event:", even_event)
print("Probability of an even number:", probability_even)
```

### Interpretation
There are 3 favorable outcomes out of 6 possible outcomes, so the probability is 3/6 = 0.5.

## 3. Sample Space

The **sample space** is the complete set of all possible outcomes of a random experiment.

It is commonly represented by `S`.

### Why is it used?
It provides the complete set of possibilities from which events are defined.

### When is it used?
It is useful when setting up a probability problem and identifying all possible outcomes.

### Real-world analogy
When tossing a coin once, the sample space is `{Head, Tail}`.

### Program:

```python
# Sample space example

sample_space = {"Head", "Tail"}

print("Sample Space:", sample_space)
print("Number of possible outcomes:", len(sample_space))
```

### Another example: rolling a die

```python
# Sample space for a die

sample_space = {1, 2, 3, 4, 5, 6}

print("Sample Space:", sample_space)
```

## 4. Independent Events

Two events are **independent** if the occurrence of one event does not change the probability of the other event.

For independent events:

P(A ∩ B) = P(A) × P(B)

### Why is it used?
It helps calculate the probability of two independent events happening together.

### When is it used?
It is used when one event has no effect on another, such as separate coin tosses.

### Real-world analogy
Tossing a fair coin twice: the result of the first toss does not affect the second toss.

### Program:

```python
# Independent events: getting Head on two separate coin tosses

p_head_first = 1 / 2
p_head_second = 1 / 2

probability_both_heads = p_head_first * p_head_second

print("P(Head on first toss):", p_head_first)
print("P(Head on second toss):", p_head_second)
print("P(Head and Head):", probability_both_heads)
```

### Interpretation
The probability of getting heads on both tosses is 1/4 = 0.25, or 25%.

## 5. Dependent Events

Two events are **dependent** when the occurrence of one event affects the probability of the other.

For dependent events:

P(A ∩ B) = P(A) × P(B|A)

### Why is it used?
It helps calculate probabilities when the first event changes the possible outcomes for the next event.

### When is it used?
It is used in situations such as drawing objects **without replacement**, where the first selection changes the remaining population.

### Real-world analogy
Suppose a bag contains 3 red and 2 blue balls. If you select one red ball and do not put it back, there are fewer red balls available for the second selection.

### Program:

```python
# Dependent events: drawing two red balls without replacement

red = 3
blue = 2
total = red + blue

p_first_red = red / total

# After drawing one red ball, one red ball is removed.
p_second_red_given_first = (red - 1) / (total - 1)

p_two_red = p_first_red * p_second_red_given_first

print("P(first red):", p_first_red)
print("P(second red | first red):", p_second_red_given_first)
print("P(two red balls):", p_two_red)
```

### Interpretation
The second probability changes after the first red ball is removed, so the events are dependent.

## 6. Conditional Probability

**Conditional probability** is the probability of an event occurring given that another event has already occurred.

P(A|B) = P(A ∩ B) / P(B)

It is read as: "Probability of A given B."

### Why is it used?
It helps us update a probability when we have additional information.

### When is it used?
It is used in medical testing, recommendation systems, spam detection, risk analysis, and machine learning.

### Real-world analogy
Suppose we want to know the probability that a student passed an exam **given that** the student attended all classes. The attendance information changes the situation we are analyzing.

### Program:

```python
# Conditional probability example

# Out of 100 customers:
# 40 are premium customers
# 30 customers made a purchase
# 20 are both premium and made a purchase

premium = 40
purchase = 30
premium_and_purchase = 20

p_purchase_given_premium = premium_and_purchase / premium

print("P(Purchase | Premium):", p_purchase_given_premium)
```

### Interpretation
The result is 0.5, meaning 50% of premium customers in this example made a purchase.

## 7. Bayes' Theorem

**Bayes' Theorem** helps us update the probability of a hypothesis when new evidence becomes available.

P(A|B) = [P(B|A) × P(A)] / P(B)

Where:
- P(A) = prior probability
- P(B|A) = likelihood
- P(B) = probability of evidence
- P(A|B) = updated/posterior probability

### Why is it used?
It helps update our belief about an event using new evidence.

### When is it used?
It is widely used in spam filtering, medical diagnosis, fraud detection, recommendation systems, and machine learning.

### Real-world analogy
Suppose a spam filter receives an email containing a word commonly found in spam. Bayes' theorem can help update the probability that the email is spam based on that evidence.

### Program:

```python
# Bayes' Theorem example: spam classification

p_spam = 0.20
p_word_given_spam = 0.80
p_word_given_not_spam = 0.10
p_not_spam = 1 - p_spam

p_word = (
    p_word_given_spam * p_spam
    + p_word_given_not_spam * p_not_spam
)

p_spam_given_word = (
    p_word_given_spam * p_spam
) / p_word

print("P(Spam):", p_spam)
print("P(Word):", p_word)
print("P(Spam | Word):", p_spam_given_word)
```

### Interpretation
The result is the updated probability that an email is spam after observing the selected word.

## 8. Probability Rules

Probability rules are basic mathematical rules used to calculate probabilities of combined events.

### Important rules

**Rule 1 – Complement**

P(Aᶜ) = 1 − P(A)

**Rule 2 – Addition rule**

For mutually exclusive events:

P(A ∪ B) = P(A) + P(B)

For general events:

P(A ∪ B) = P(A) + P(B) − P(A ∩ B)

**Rule 3 – Multiplication rule**

For independent events:

P(A ∩ B) = P(A)P(B)

For dependent events:

P(A ∩ B) = P(A)P(B|A)

### Why is it used?
These rules provide the foundation for solving probability problems.

### When is it used?
They are used whenever multiple events, complements, intersections, or unions need to be analyzed.

### Real-world analogy
If you want the probability of a customer buying either Product A or Product B, you can use the addition rule while accounting for customers who buy both.

```python
# Probability rules example

p_A = 0.6
p_B = 0.5
p_A_and_B = 0.3

# Complement
p_not_A = 1 - p_A

# General addition rule
p_A_or_B = p_A + p_B - p_A_and_B

# Multiplication rule relationship
p_B_given_A = p_A_and_B / p_A

print("P(not A):", p_not_A)
print("P(A or B):", p_A_or_B)
print("P(B | A):", p_B_given_A)
```

### Interpretation
The addition rule avoids double-counting the outcomes where both A and B happen.

## 9. Random Variables

A **random variable** is a variable that assigns a numerical value to the outcome of a random experiment.

Two common types:
- **Discrete random variable:** takes countable values, such as number of heads.
- **Continuous random variable:** can take values within a range, such as height or temperature.

### Why is it used?
Random variables convert uncertain outcomes into numerical values that can be analyzed mathematically.

### When is it used?
They are used throughout probability, statistics, data science, and machine learning.

### Real-world analogy
If you toss three coins, the random variable X could represent the number of heads. X can be 0, 1, 2, or 3.

### Program:

```python
# Discrete random variable example

import random

random.seed(42)

tosses = [random.choice(["H", "T"]) for _ in range(3)]
number_of_heads = tosses.count("H")

print("Coin tosses:", tosses)
print("Random variable X (number of heads):", number_of_heads)
```

### Continuous random variable example

```python
# Continuous random variable example

import random

temperature = random.uniform(20, 35)

print("Random temperature:", temperature)
```

### Interpretation
The number of heads is a discrete random variable, while the generated temperature can take many possible values within the selected interval.

## 10. Expected Value

**Expected value** is the probability-weighted average of all possible outcomes.

For a discrete random variable:

E(X) = Σ xP(X = x)

### Why is it used?
It tells us the long-run average outcome we would expect if the same random process were repeated many times.

### When is it used?
It is used in decision-making under uncertainty, finance, games, insurance, risk analysis, and machine learning.

### Real-world analogy
Suppose a game gives ₹100 with probability 0.2 and ₹0 with probability 0.8. The expected value is ₹20. This does not mean every game gives ₹20; it represents the long-run average.

### Program:

```python
# Expected value example

outcomes = [100, 0]
probabilities = [0.2, 0.8]

expected_value = sum(
    outcome * probability
    for outcome, probability in zip(outcomes, probabilities)
)

print("Outcomes:", outcomes)
print("Probabilities:", probabilities)
print("Expected Value:", expected_value)
```

### Interpretation
The expected value is ₹20. Over a very large number of repetitions, the average payout would tend toward ₹20 under the stated probabilities.
