## 1. Normal Distribution

A **normal distribution** is a continuous probability distribution that has a bell-shaped curve. Most observations are concentrated around the mean, and the values become less common as we move away from the mean.

In a perfectly normal distribution:
- Mean = Median = Mode
- The distribution is symmetric around the mean.

### Why is it used?
It helps us understand and model data that naturally clusters around an average value.

### When is it used?
It is commonly used in statistics, quality control, measurement analysis, and many machine-learning/statistical methods.

### Real-world analogy
Imagine the heights of a large group of adults. Most people may have heights around the average, while very short and very tall people are less common.

### Program:

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate normally distributed data
np.random.seed(42)
data = np.random.normal(loc=170, scale=8, size=1000)

plt.hist(data, bins=30, density=True)
plt.title("Normal Distribution of Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Density")
plt.show()

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

### Interpretation
The histogram should look approximately bell-shaped, with most values concentrated around the mean.

## 2. Uniform Distribution

A **uniform distribution** is a distribution where values within a specified interval have equal probability density.

For a continuous uniform distribution between `a` and `b`, values between `a` and `b` are equally likely in terms of probability density.

### Why is it used?
It is useful for representing situations where there is no preference for one value over another within a defined range.

### When is it used?
It is used in simulations, random sampling, probability experiments, and computer algorithms.

### Real-world analogy
Imagine selecting a random point on a ruler between 0 cm and 10 cm. Every location within that interval is equally likely to be selected.

### Program:

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.uniform(low=0, high=10, size=1000)

plt.hist(data, bins=20, density=True)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()

print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Mean:", np.mean(data))
```

### Interpretation
The histogram should appear relatively flat because the values are generated uniformly between 0 and 10.

## 3. Gaussian Distribution

A **Gaussian distribution** is another name commonly used for the **normal distribution**. It is a bell-shaped, symmetric distribution described by its mean and standard deviation.

So, for this notebook:
**Gaussian distribution ≈ Normal distribution.**

### Why is it used?
It provides a mathematical model for continuous data that is approximately bell-shaped.

### When is it used?
It is widely used in probability, statistics, signal processing, measurement modeling, and machine learning.

### Real-world analogy
Think about repeated measurement of the same object. Small measurement differences may occur around the true average, with larger deviations being less common.

### Program

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 500)

mean = 0
std = 1

pdf = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std) ** 2
)

plt.plot(x, pdf)
plt.title("Gaussian (Normal) Distribution")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.show()
```

### Interpretation
The curve reaches its highest point around the mean and decreases symmetrically as the distance from the mean increases.

## 4. Left Skew (Negative Skew)

A **left-skewed distribution** has a longer tail on the left side. A small number of unusually low values pull the distribution toward the left.

Typically:
**Mean < Median**  
when the distribution is clearly negatively skewed.

### Why is it used?
It helps us recognize that data is not evenly distributed and that unusually low values may be affecting the average.

### When is it used?
It is useful when analyzing income, test results, response times, measurements, or any dataset where a few low observations create a long left tail.

### Real-world analogy
Suppose most employees have high performance scores, but a few employees have very low scores. Those few low scores can create a longer tail on the left.

### program

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Create a negatively skewed dataset
data = 100 - np.random.exponential(scale=10, size=1000)

plt.hist(data, bins=30)
plt.title("Left-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(data))
print("Median:", np.median(data))
```

### Interpretation
The distribution has a longer tail toward lower values. The mean can be pulled toward that lower tail.

## 5. Right Skew (Positive Skew)

A **right-skewed distribution** has a longer tail on the right side. A small number of unusually high values pull the distribution toward the right.

Typically:
**Mean > Median**  
when the distribution is clearly positively skewed.

### Why is it used?
It helps identify datasets where a few very large values affect the average.

### When is it used?
It is common in income, house prices, waiting times, sales amounts, and transaction values.

### Real-world analogy
Consider salaries in a company. Most employees may have salaries within a normal range, while a small number of senior executives earn much more, creating a long right tail.

### program

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.exponential(scale=10, size=1000)

plt.hist(data, bins=30)
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(data))
print("Median:", np.median(data))
```

### Interpretation
The distribution has a longer tail toward higher values. The mean can be pulled toward the high-value tail.

## 6. Symmetric Distribution

A **symmetric distribution** has a similar shape on both sides of its center. The left and right sides are mirror-like around the central point.

A normal distribution is an important example of a symmetric distribution.

### Why is it used?
It helps us understand whether observations are balanced around a central value.

### When is it used?
It is useful when describing the shape of a dataset and checking whether methods that assume symmetry may be appropriate.

### Real-world analogy
Imagine a perfectly balanced seesaw. The amount of data on the left and right sides of the center is distributed in a similar way.

### program

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=1000)

plt.hist(data, bins=30)
plt.title("Approximately Symmetric Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(data))
print("Median:", np.median(data))
```

### Interpretation
For a sufficiently large sample from a normal distribution, the mean and median should be close, and the two sides of the histogram should look approximately balanced.

## 7. Kurtosis

**Kurtosis** is a statistical measure related to the shape of a distribution, particularly the weight of its tails and the concentration of observations around the center.

A common definition used in Python's SciPy is **excess kurtosis**, where a normal distribution has kurtosis approximately equal to 0.

Types commonly discussed:
- **Mesokurtic:** approximately normal tail behavior; excess kurtosis ≈ 0
- **Leptokurtic:** heavier tails; excess kurtosis > 0
- **Platykurtic:** lighter tails; excess kurtosis < 0

### Why is it used?
It helps identify whether a dataset has relatively heavy or light tails compared with a normal distribution.

### When is it used?
It is useful in risk analysis, financial data, quality control, and statistical distribution analysis where extreme observations matter.

### Real-world analogy
Imagine two groups with the same average and similar spread. One group has more unusually extreme observations. Kurtosis can help describe this difference in tail behavior.

### program

```python
import numpy as np
from scipy.stats import kurtosis

np.random.seed(42)

data = np.random.normal(loc=50, scale=10, size=1000)

excess_kurtosis = kurtosis(data)

print("Excess Kurtosis:", excess_kurtosis)
```

### Interpretation
For a large sample generated from a normal distribution, the excess kurtosis should be close to 0, although the exact sample value will vary because of random sampling.

## 8. Empirical Rule (68–95–99.7 Rule)

The **Empirical Rule** describes the approximate percentage of observations within certain standard-deviation ranges of the mean for a normal distribution:

- About **68%** are within ±1 standard deviation
- About **95%** are within ±2 standard deviations
- About **99.7%** are within ±3 standard deviations

This rule applies to data that follows a normal distribution approximately.

### Why is it used?
It provides a quick way to estimate how much data should fall within a certain distance from the mean.

### When is it used?
It is useful for understanding normal data, checking unusual observations, and making quick statistical estimates.

### Real-world analogy
Suppose exam scores are approximately normally distributed with an average of 70 and a standard deviation of 10. The Empirical Rule suggests approximately:
- 68% of scores are between 60 and 80
- 95% are between 50 and 90
- 99.7% are between 40 and 100
### program

```python
# Empirical Rule example

mean = 70
std = 10

within_1 = (mean - std, mean + std)
within_2 = (mean - 2 * std, mean + 2 * std)
within_3 = (mean - 3 * std, mean + 3 * std)

print("Mean:", mean)
print("Standard Deviation:", std)
print("Approximately 68%:", within_1)
print("Approximately 95%:", within_2)
print("Approximately 99.7%:", within_3)
```

### Verify the Empirical Rule with simulated data

```python
import numpy as np

np.random.seed(42)

data = np.random.normal(loc=70, scale=10, size=100000)

mean = np.mean(data)
std = np.std(data)

percentage_1 = np.mean((data >= mean - std) & (data <= mean + std)) * 100
percentage_2 = np.mean((data >= mean - 2 * std) & (data <= mean + 2 * std)) * 100
percentage_3 = np.mean((data >= mean - 3 * std) & (data <= mean + 3 * std)) * 100

print(f"Within ±1 SD: {percentage_1:.2f}%")
print(f"Within ±2 SD: {percentage_2:.2f}%")
print(f"Within ±3 SD: {percentage_3:.2f}%")
```

### Interpretation
The simulated percentages should be close to 68%, 95%, and 99.7%. They will not be exactly equal because the dataset is a random sample.

**Important:** The Empirical Rule should not be applied blindly to strongly skewed or non-normal data.
