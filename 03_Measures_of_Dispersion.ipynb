# Notebook 3 – Measures of Dispersion

## Objective
Measures of dispersion tell us **how spread out or varied the data is**.

In this notebook, we will learn:
- Range
- Variance
- Standard Deviation
- Quartiles
- Percentiles
- Interquartile Range (IQR)
- Coefficient of Variation

## 1. Range

**Range** is the difference between the largest value and the smallest value in a dataset.

**Formula:**

Range = Maximum value − Minimum value

### Why is it used?
Range gives a quick idea of the total spread of the data.

### When is it used?
It is useful when we want a simple and quick measure of variation, especially for small datasets.

### Real-world analogy
Suppose the daily temperatures in a week range from 24°C to 32°C. The range tells us how much the temperature varied during that week.

### Example
Data: 10, 15, 20, 25, 30

Range = 30 − 10 = 20

```python
# Range example
data = [10, 15, 20, 25, 30]

data_range = max(data) - min(data)

print("Data:", data)
print("Maximum:", max(data))
print("Minimum:", min(data))
print("Range:", data_range)
```

### Interpretation
A range of **20** means the distance between the smallest and largest values is 20 units.

## 2. Variance

**Variance** measures how far the values in a dataset are spread from the mean, using squared differences.

For a population:

Variance = Σ(x − μ)² / N

### Why is it used?
It helps us understand the amount of variability in the data.

### When is it used?
It is used when analyzing the overall variation in data and as a foundation for other statistical methods, including standard deviation and many machine-learning techniques.

### Real-world analogy
Imagine students standing around their average height. If most students stand close to the average, variance is small. If they are spread far away, variance is large.

### Example
Data: 10, 20, 30

Mean = 20

Squared deviations:
- (10 − 20)² = 100
- (20 − 20)² = 0
- (30 − 20)² = 100

Population variance = (100 + 0 + 100) / 3 = 66.67

```python
# Variance example
data = [10, 20, 30]

mean = sum(data) / len(data)
squared_deviations = [(x - mean) ** 2 for x in data]
population_variance = sum(squared_deviations) / len(data)

print("Data:", data)
print("Mean:", mean)
print("Squared deviations:", squared_deviations)
print("Population Variance:", population_variance)
```

### Python using NumPy
`np.var()` calculates population variance by default.

```python
import numpy as np

data = [10, 20, 30]

print("Population Variance:", np.var(data))
print("Sample Variance:", np.var(data, ddof=1))
```

### Interpretation
A larger variance means the observations are more spread out around the mean. A smaller variance means they are closer to the mean.

## 3. Standard Deviation

**Standard deviation** measures the typical spread of values around the mean. It is the square root of variance.

Standard deviation = √Variance

### Why is it used?
It gives a measure of spread in the **same unit as the original data**, making it easier to interpret than variance.

### When is it used?
It is commonly used to understand consistency, variability, and risk in data.

### Real-world analogy
If the average delivery time is 30 minutes and the standard deviation is small, delivery times are usually close to 30 minutes. A large standard deviation means delivery times vary more.

### program

```python
# Standard deviation example
import math

data = [10, 20, 30]

mean = sum(data) / len(data)
variance = sum((x - mean) ** 2 for x in data) / len(data)
standard_deviation = math.sqrt(variance)

print("Data:", data)
print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)
```

### Using NumPy

```python
import numpy as np

data = [10, 20, 30]

print("Population Standard Deviation:", np.std(data))
print("Sample Standard Deviation:", np.std(data, ddof=1))
```

### Interpretation
A smaller standard deviation indicates that values are generally closer to the mean. A larger standard deviation indicates greater spread.

## 4. Quartiles

**Quartiles** divide ordered data into four parts.

- **Q1 (First Quartile):** approximately 25% of observations are below it.
- **Q2 (Second Quartile):** the median, approximately 50% of observations are below it.
- **Q3 (Third Quartile):** approximately 75% of observations are below it.

### Why is it used?
Quartiles help us understand the distribution of data without depending heavily on extreme values.

### When is it used?
They are useful for analyzing distributions, identifying spread, and detecting possible outliers.

### Real-world analogy
Imagine arranging students by exam marks from lowest to highest and dividing them into four sections. Q1, Q2, and Q3 mark the boundaries between those sections.

### program

```python
# Quartiles example
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 30]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Data:", data)
print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

### Interpretation
Quartiles show where the lower 25%, middle 50%, and upper 25% portions of the data lie.

## 5. Percentiles

A **percentile** indicates the value below which a particular percentage of observations fall.

For example, the 90th percentile is the value below which approximately 90% of the observations fall.

### Why is it used?
Percentiles help compare an observation with the rest of a dataset.

### When is it used?
They are commonly used in exam scores, employee performance analysis, salary analysis, customer metrics, and machine-learning data analysis.

### Real-world analogy
If a student's score is at the 90th percentile, it means the student scored higher than approximately 90% of the students in the reference dataset.

### program

```python
# Percentile example
import numpy as np

scores = [45, 50, 55, 60, 65, 70, 75, 80, 85, 95]

p25 = np.percentile(scores, 25)
p50 = np.percentile(scores, 50)
p90 = np.percentile(scores, 90)

print("Scores:", scores)
print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("90th Percentile:", p90)
```

### Interpretation
Percentiles are useful for understanding the relative position of a value within a dataset. The exact percentile calculation can vary slightly between statistical software because different interpolation methods may be used.

## 6. Interquartile Range (IQR)

**Interquartile Range (IQR)** is the difference between the third quartile and first quartile.

IQR = Q3 − Q1

It represents the spread of the **middle 50%** of the data.

### Why is it used?
IQR measures spread while reducing the influence of extreme values and is useful for identifying potential outliers.

### When is it used?
It is especially useful when the dataset contains outliers or is skewed.

### Real-world analogy
Suppose salaries contain a few extremely high earners. The IQR focuses on the middle 50% of salaries instead of being strongly affected by those extreme salaries.

### program

```python
# IQR example
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 30, 100]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

print("Data:", data)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

### Detecting potential outliers using the IQR rule

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds are commonly treated as **potential outliers**.

```python
# IQR outlier detection example
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 30, 100]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:", outliers)
```

### Interpretation
IQR tells us how spread out the middle 50% of observations are. A larger IQR means more spread within the middle portion of the data.

## 7. Coefficient of Variation (CV)

**Coefficient of Variation (CV)** measures the standard deviation relative to the mean.

CV = (Standard Deviation / Mean) × 100

### Why is it used?
CV allows us to compare variability between datasets even when their means or units are different.

### When is it used?
It is useful when comparing the relative consistency or variation of different datasets, such as investment returns, production processes, or measurements.

### Real-world analogy
Suppose two machines produce products with different average sizes. Comparing only their standard deviations may not be fair because their average sizes are different. CV expresses variation relative to the average.

### program

```python
# Coefficient of Variation example
import numpy as np

data = [90, 100, 110, 100, 100]

mean = np.mean(data)
std = np.std(data)

cv = (std / mean) * 100

print("Data:", data)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Coefficient of Variation:", cv, "%")
```

### Interpretation
A lower CV indicates lower relative variability, while a higher CV indicates higher relative variability.

**Important:** CV is generally most meaningful when the mean is positive and not close to zero.
