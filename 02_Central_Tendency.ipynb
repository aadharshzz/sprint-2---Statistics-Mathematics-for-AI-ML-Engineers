Measures of central tendency help us find a **typical or central value** in a dataset.

## 1. Mean

### Definition
The **mean**, commonly called the average, is the sum of all values divided by the number of values.

### What is it?
It gives the arithmetic average of a dataset.

### Why do we use it?
To represent many numerical observations with one central value.

### When do we use it?
Use it when observations should have equal importance and extreme outliers are not strongly distorting the result.

### Formula
**Mean = Sum of values / Number of values**

`Mean = (x₁ + x₂ + ... + xₙ) / n`

### Numerical Example
Data: `10, 20, 30, 40, 50`

Mean = `(10 + 20 + 30 + 40 + 50) / 5 = 30`

### Real-world analogy
Five friends spend ₹100, ₹200, ₹300, ₹400, and ₹500 on a trip. The mean tells us their average spending.

### Business Example
A company calculates average monthly sales to understand its typical sales level.

### AI/ML Example
An ML engineer may calculate the average of a numerical feature during exploratory data analysis.

### Interpretation
A mean of **30** means the arithmetic average of all observations is 30.

```python
# Real-world analogy: average spending of five friends
spending = [100, 200, 300, 400, 500]

mean_spending = sum(spending) / len(spending)

print("Spending:", spending)
print("Mean spending: ₹", mean_spending)
```

```python
import statistics

marks = [65, 70, 75, 80, 90]
print("Mean marks:", statistics.mean(marks))
```

## 2. Median

### Definition
The **median** is the middle value after arranging observations in order.

### What is it?
It divides an ordered dataset into two halves.

### Why do we use it?
It is less affected by extreme values than the mean.

### When do we use it?
Use it when data is skewed or contains outliers.

### Formula
After sorting:
- Odd number of values → middle value
- Even number of values → average of the two middle values

### Numerical Example
`10, 20, 30, 40, 1000`

Median = **30**

### Real-world analogy
House prices are ₹10 lakh, ₹20 lakh, ₹30 lakh, ₹40 lakh and ₹1 crore. The median is ₹30 lakh, which can describe the middle house better than the mean.

### Business Example
Median salary can be useful when a few very highly paid employees make the average salary unusually high.

### AI/ML Example
Median is useful when exploring features such as income or house price that contain outliers.

### Interpretation
A median of 30 means half the observations are at or below 30 and half are at or above 30.

```python
# Real-world analogy: house prices (values in lakh rupees)
import statistics

house_prices = [10, 20, 30, 40, 1000]

print("House prices:", house_prices)
print("Median price:", statistics.median(house_prices), "lakh")
```

```python
# Mean vs median when an outlier is present
salaries = [20000, 22000, 25000, 28000, 200000]

print("Mean salary:", statistics.mean(salaries))
print("Median salary:", statistics.median(salaries))
```

## 3. Mode

### Definition
The **mode** is the value or category that occurs most frequently.

### What is it?
It identifies the most common observation.

### Why do we use it?
To find the most popular or frequently occurring value/category.

### When do we use it?
Especially useful for categorical data, customer preferences, product sizes, colors, etc.

### Formula
There is no arithmetic formula.

**Mode = Most frequently occurring value**

### Numerical Example
`10, 20, 20, 30, 40`

Mode = **20**

### Real-world analogy
A clothing store records sizes: `M, L, M, S, M, L`. The most commonly sold size is M.

### Business Example
A retailer can find its most popular shirt size or product color.

### AI/ML Example
Mode can help understand categorical features and is commonly used as a simple method for filling missing categorical values.

### Interpretation
If the mode is M, M is the most frequently occurring category in the observed data.

```python
# Real-world analogy: shirt sizes sold
import statistics

sizes = ["M", "L", "M", "S", "M", "L"]

print("Sizes sold:", sizes)
print("Most common size (mode):", statistics.mode(sizes))
```

## 4. Weighted Mean

### Definition
The **weighted mean** is an average where different values have different levels of importance.

### What is it?
Some observations contribute more than others because they have larger weights.

### Why do we use it?
A simple average may be inappropriate when all values do not have equal importance.

### When do we use it?
Examples include:
- college subjects with different credits
- investment returns with different investment amounts
- business KPIs with different importance

### Formula
**Weighted Mean = Σ(value × weight) / Σ(weight)**

### Numerical Example
Marks:
- Maths = 80, weight = 3
- Science = 70, weight = 2
- English = 90, weight = 1

Weighted mean = `(80×3 + 70×2 + 90×1) / 6 = 76.67`

### Real-world analogy
If a final exam counts more than a small assignment, the final exam should have a larger weight.

### Business Example
A company can calculate an overall performance score by weighting sales, quality, and customer satisfaction differently.

### AI/ML Example
Weighted averages can be used when observations or evaluation components have different importance.

### Interpretation
A weighted mean of 76.67 means the overall score after considering the importance of each subject is approximately 76.67.

```python
# Real-world analogy: college subjects with different credits
marks = [80, 70, 90]
credits = [3, 2, 1]

weighted_mean = sum(m * w for m, w in zip(marks, credits)) / sum(credits)

print("Marks:", marks)
print("Credits/weights:", credits)
print("Weighted mean:", round(weighted_mean, 2))
```

## 5. Geometric Mean

### Definition
The **geometric mean** is the nth root of the product of n positive values.

### What is it?
It is an average based on multiplication rather than addition.

### Why do we use it?
It is useful for multiplicative changes, growth rates, ratios, and compounded values.

### When do we use it?
Common uses:
- investment growth
- population growth
- percentage growth over several periods
- multiplicative ratios

It is generally used with **positive values**.

### Formula
**Geometric Mean = (x₁ × x₂ × ... × xₙ)^(1/n)**

### Numerical Example
Growth factors: `1.10, 1.20, 1.30`

Geometric mean = `(1.10 × 1.20 × 1.30)^(1/3) ≈ 1.196`

Equivalent average growth rate ≈ **19.6%**.

### Real-world analogy
An investment grows by different percentages each year. Because each year's growth builds on the previous year's value, multiplication is involved.

### Business Example
A company can use geometric mean to summarize multi-year compounded growth.

### AI/ML Example
Geometric means can be useful when aggregating multiplicative ratios or growth factors in appropriate analytical problems.

### Interpretation
A geometric mean of about 1.196 represents an equivalent constant growth factor of about 1.196 per period.

```python
# Real-world analogy: investment growth
growth_factors = [1.10, 1.20, 1.30]

product = 1
for factor in growth_factors:
    product *= factor

geometric_mean = product ** (1 / len(growth_factors))

print("Growth factors:", growth_factors)
print("Geometric mean:", round(geometric_mean, 4))
print("Equivalent average growth rate:", round((geometric_mean - 1) * 100, 2), "%")
```

```python
# Geometric mean using Python's math module
import math

values = [2, 8, 32]
geometric_mean = math.prod(values) ** (1 / len(values))

print("Values:", values)
print("Geometric mean:", geometric_mean)
```

## 6. Harmonic Mean

### Definition
The **harmonic mean** is the reciprocal of the arithmetic mean of the reciprocals.

### What is it?
It is an average that gives greater influence to smaller values.

### Why do we use it?
It is useful for suitable rates and ratios, especially when the same quantity is involved across different rates.

### When do we use it?
Examples:
- average speed over equal distances
- rates
- ratios
- ML metrics such as F1-score

### Formula
**Harmonic Mean = n / (1/x₁ + 1/x₂ + ... + 1/xₙ)**

### Numerical Example
A vehicle travels equal distances at 60 km/h and 40 km/h.

Harmonic mean = `2 / (1/60 + 1/40) = 48 km/h`

### Real-world analogy
Driving the same distance to a destination at 60 km/h and returning at 40 km/h gives an overall average speed of 48 km/h.

### Business Example
Harmonic means can be useful for combining appropriate rate-based measures.

### AI/ML Example
The F1-score is the harmonic mean of precision and recall:

`F1 = 2 × Precision × Recall / (Precision + Recall)`

### Interpretation
A harmonic mean of 48 km/h is the correct average speed for equal-distance trips at 60 km/h and 40 km/h.

```python
# Real-world analogy: average speed for equal distances
speed_1 = 60
speed_2 = 40

harmonic_mean_speed = 2 / ((1 / speed_1) + (1 / speed_2))

print("First speed:", speed_1, "km/h")
print("Return speed:", speed_2, "km/h")
print("Harmonic mean speed:", harmonic_mean_speed, "km/h")
```

```python
# AI/ML example: F1-score
precision = 0.80
recall = 0.60

f1_score = 2 * precision * recall / (precision + recall)

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", round(f1_score, 2))
```
