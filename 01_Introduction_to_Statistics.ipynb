## 1. What is Statistics?

### What is it?
**Statistics** is the process of collecting, organizing, analyzing, interpreting, and presenting data.

### Why do we use it?
We use statistics to turn raw data into useful information and make better decisions.

### When do we use it?
Whenever we have data and want to understand patterns, differences, averages, or trends.

### Real-world analogy
Suppose a teacher has marks of 50 students. Looking at all 50 marks one by one is difficult. Statistics can help the teacher find the average mark, highest mark, lowest mark, and overall performance.

### Python example
We calculate the average marks of students.

```python
marks = [72, 85, 60, 90, 78]

average = sum(marks) / len(marks)

print("Student marks:", marks)
print("Average mark:", average)
```

## 2. Why Statistics is Important in AI/ML

### What is it?
Statistics provides methods for understanding data before and after building an AI/ML model.

### Why do we use it?
Statistics helps us:
- understand datasets
- find patterns
- detect unusual values
- summarize data
- choose useful features
- evaluate model results
- make decisions from data

### When do we use it?
Almost throughout an ML workflow: especially during data exploration, preprocessing, analysis, and evaluation.

### Real-world analogy
Imagine a doctor checking a patient's temperature, blood pressure, and other measurements before deciding what action to take. Similarly, an ML engineer examines data before training a model.

### Python example
We inspect simple student data before using it for analysis.

```python
students = {
    "marks": [72, 85, 60, 90, 78],
    "attendance": [80, 95, 70, 98, 88]
}

print("Average marks:", sum(students["marks"]) / len(students["marks"]))
print("Average attendance:", sum(students["attendance"]) / len(students["attendance"]))
```

## 3. Statistics vs Machine Learning

| Statistics | Machine Learning |
|---|---|
| Focuses on understanding and describing data | Focuses on learning patterns to make predictions or decisions |
| Often emphasizes inference and relationships | Often emphasizes prediction/generalization |
| Uses mathematical and statistical methods | Uses statistical, mathematical, and computational methods |
| Example: finding average salary | Example: predicting future salary |

### Real-world analogy
A shop owner studies previous sales to understand which products sell well — this is statistical analysis.

If the shop uses previous sales to automatically predict next month's sales, that is a machine learning task.

### Python example
First we summarize historical sales. Then we make a very simple prediction using the average as a baseline.

```python
sales = [100, 120, 90, 150, 140]

average_sales = sum(sales) / len(sales)

print("Historical sales:", sales)
print("Average sales:", average_sales)
print("Simple baseline prediction for next month:", average_sales)
```

## 4. Descriptive Statistics

### What is it?
**Descriptive statistics** summarizes the data we already have.

Common examples:
- Mean
- Median
- Mode
- Minimum
- Maximum
- Range

### Why do we use it?
To quickly understand the main characteristics of a dataset.

### When do we use it?
During data exploration and reporting, before deeper analysis or ML modeling.

### Real-world analogy
A teacher says, "The class average is 75, the highest mark is 95, and the lowest is 48." The teacher is describing the existing class data.

### Python example

```python
import statistics

marks = [60, 70, 70, 80, 90]

print("Mean:", statistics.mean(marks))
print("Median:", statistics.median(marks))
print("Mode:", statistics.mode(marks))
print("Minimum:", min(marks))
print("Maximum:", max(marks))
print("Range:", max(marks) - min(marks))
```

## 5. Inferential Statistics

### What is it?
**Inferential statistics** uses data from a sample to make conclusions or estimates about a larger population.

### Why do we use it?
Studying every member of a population can be expensive or time-consuming. A carefully selected sample can provide useful information about the population.

### When do we use it?
When we want to estimate, test, or generalize findings from a sample to a population.

### Real-world analogy
A company wants to know whether customers like a new product. Instead of asking every customer, it surveys a sample of customers and uses the results to understand the larger customer population.

### Python example
We use a sample of customer ratings to estimate the average rating.

```python
sample_ratings = [4, 5, 4, 3, 5, 4, 5, 4]

sample_average = sum(sample_ratings) / len(sample_ratings)

print("Sample ratings:", sample_ratings)
print("Sample average rating:", sample_average)
print("This sample average can be used as an estimate of the larger customer group's average rating.")
```

## 6. Population

### What is it?
A **population** is the complete group of people, objects, or observations we are interested in studying.

### Why do we use it?
It defines the full group about which we want information.

### When do we use it?
Whenever we define the scope of a statistical study.

### Real-world analogy
All students in a college are the population if we want to study the college's student performance.

### Python example

```python
college_students = [
    {"name": "Aadharsh", "marks": 75},
    {"name": "Vedha", "marks": 82},
    {"name": "Yogesh", "marks": 68},
    {"name": "Vishal", "marks": 91},
    {"name": "Hemnath", "marks": 78}
]

print("Population size:", len(college_students))
print("Population:", college_students)
```

## 7. Sample

### What is it?
A **sample** is a smaller group selected from a population.

### Why do we use it?
It is often faster, cheaper, and easier to study a sample than the entire population.

### When do we use it?
When the population is large or collecting data from everyone is impractical.

### Real-world analogy
If a college has 10,000 students, selecting 500 students for a survey creates a sample.

### Python example

```python
population_marks = [55, 60, 65, 70, 75, 80, 85, 90, 95, 100]

sample_marks = population_marks[::2]

print("Population:", population_marks)
print("Sample:", sample_marks)
```

## 8. Census

### What is it?
A **census** collects information from every member of the population.

### Why do we use it?
To obtain complete information about the entire population.

### When do we use it?
When collecting data from everyone is possible and the complete information is needed.

### Real-world analogy
A national population census attempts to collect information about all people in the country.

### Python example

```python
population = ["A", "B", "C", "D", "E"]

# A census studies every member.
for person in population:
    print("Collecting information from:", person)
```

## 9. Sampling

### What is it?
**Sampling** is the process of selecting a sample from a population.

### Why do we use it?
To reduce the time, cost, and effort needed to collect data.

### When do we use it?
When the population is too large to study completely.

### Real-world analogy
A supermarket wants feedback from 100 customers instead of interviewing every customer who visits the store.

### Python example

```python
customers = ["Customer1", "Customer2", "Customer3", "Customer4", "Customer5",
             "Customer6", "Customer7", "Customer8", "Customer9", "Customer10"]

sample = customers[:4]

print("Population:", customers)
print("Selected sample:", sample)
```

## 10. Variables

### What is it?
A **variable** is a characteristic or value that can change from one observation to another.

Examples:
- Age
- Salary
- City
- Marks
- Gender

### Why do we use it?
Variables describe the properties of the observations we collect.

### When do we use it?
Whenever we collect or analyze data.

### Real-world analogy
For an employee, age, department, and salary can all be variables.

### Python example

```python
employee_age = 25
employee_salary = 45000
employee_city = "Coimbatore"

print("Age:", employee_age)
print("Salary:", employee_salary)
print("City:", employee_city)
```

## 11. Features

### What is it?
A **feature** is an input variable used by a machine learning model to learn patterns and make predictions.

### Why do we use it?
Features provide information that the model uses to predict a target/output.

### When do we use it?
During ML model training and prediction.

### Real-world analogy
To predict a house price, features could include area, number of bedrooms, and age of the house. The price is the target.

### Python example

```python
house = {
    "area_sqft": 1200,
    "bedrooms": 3,
    "age_years": 5,
    "price": 7500000
}

features = [house["area_sqft"], house["bedrooms"], house["age_years"]]
target = house["price"]

print("Features:", features)
print("Target:", target)
```

## 12. Observation

### What is it?
An **observation** is one individual record or data point in a dataset.

### Why do we use it?
It represents one entity, event, or measurement being studied.

### When do we use it?
Whenever we work with rows/records in a dataset.

### Real-world analogy
One student's complete information — name, age, and marks — is one observation.

### Python example

```python
students = [
    {"name": "Aadharsh", "age": 22, "marks": 78},
    {"name": "yogesh", "age": 22, "marks": 85},
    {"name": "vishal", "age": 22, "marks": 92}
]

observation = students[0]

print("One observation:", observation)
```

## 13. Dataset

### What is it?
A **dataset** is a collection of related data organized for analysis or use.

### Why do we use it?
Datasets provide the information needed for statistical analysis and machine learning.

### When do we use it?
Whenever we analyze data or train/test ML models.

### Real-world analogy
A spreadsheet containing thousands of customer records is a dataset.

### Python example

```python
dataset = [
    {"name": "Aadharsh", "age": 22, "marks": 78},
    {"name": "Yogesh", "age": 22, "marks": 85},
    {"name": "Vishal", "age": 22, "marks": 92}
]

print("Number of observations:", len(dataset))
print("First observation:", dataset[0])
```

## 14. Quantitative Data

### What is it?
**Quantitative data** is numerical data that represents amounts or measurements.

Examples:
- Age = 21
- Salary = ₹45,000
- Height = 172.5 cm
- Marks = 85

### Why do we use it?
Numbers can be measured, compared, and analyzed mathematically.

### When do we use it?
When the information represents a measurable quantity.

### Real-world analogy
A shop records the number of products sold each day. Those counts are quantitative data.

### Python example

```python
daily_sales = [25, 30, 18, 42, 35]

print("Daily sales:", daily_sales)
print("Total products sold:", sum(daily_sales))
print("Average products sold:", sum(daily_sales) / len(daily_sales))
```

## 15. Qualitative Data

### What is it?
**Qualitative data** describes categories or qualities rather than numerical amounts.

Examples:
- City = Coimbatore
- Department = HR
- Color = Blue
- Product type = Laptop

### Why do we use it?
It helps us understand categories and characteristics.

### When do we use it?
When the information is descriptive or categorical.

### Real-world analogy
A restaurant records whether customers prefer South Indian, North Indian, or Chinese food. These are categories.

### Python example

```python
food_preferences = ["South Indian", "Chinese", "South Indian", "North Indian", "Chinese"]

print("Customer preferences:", food_preferences)

from collections import Counter
print("Preference counts:", Counter(food_preferences))
```

## 16. Continuous Data

### What is it?
**Continuous data** can take any value within a range and is usually obtained by measurement.

Examples:
- Height = 172.5 cm
- Weight = 68.7 kg
- Temperature = 36.6°C
- Time = 12.45 seconds

### Why do we use it?
It represents measurements where values can have decimals.

### When do we use it?
When measuring physical quantities or other values that can vary continuously.

### Real-world analogy
A person's height can be 170 cm, 170.5 cm, 170.55 cm, and so on.

### Python example

```python
temperatures = [36.5, 36.8, 37.1, 36.6, 36.9]

print("Temperatures:", temperatures)
print("Average temperature:", sum(temperatures) / len(temperatures))
```

## 17. Discrete Data

### What is it?
**Discrete data** consists of countable values, usually whole numbers.

Examples:
- Number of students = 40
- Number of cars = 12
- Number of orders = 25

### Why do we use it?
It represents things that can be counted.

### When do we use it?
When recording counts rather than measurements.

### Real-world analogy
A parking lot can have 20 cars or 21 cars, but normally not 20.5 cars.

### Python example

```python
orders_per_day = [12, 18, 15, 22, 20]

print("Orders per day:", orders_per_day)
print("Total orders:", sum(orders_per_day))
```

## 18. Structured Data

### What is it?
**Structured data** is organized in a fixed format, commonly rows and columns.

Examples:
- Excel spreadsheets
- SQL tables
- CSV files

### Why do we use it?
It is easy to store, search, filter, and analyze using standard tools.

### When do we use it?
When data naturally fits into rows and columns.

### Real-world analogy
An employee table with columns such as employee ID, name, department, and salary is structured data.

### Python example

```python
employees = [
    {"id": 101, "name": "Aadharsh", "department": "IT", "salary": 45000},
    {"id": 102, "name": "yogesh", "department": "HR", "salary": 40000},
    {"id": 103, "name": "Vishal", "department": "IT", "salary": 50000}
]

for employee in employees:
    print(employee["id"], employee["name"], employee["department"], employee["salary"])
```

## 19. Unstructured Data

### What is it?
**Unstructured data** does not follow a fixed row-and-column structure.

Examples:
- Images
- Videos
- Audio
- Free-form text
- Social media posts

### Why do we use it?
A large amount of real-world information is unstructured. AI/ML techniques can process it to extract useful patterns.

### When do we use it?
When working with data such as images, speech, videos, or natural language.

### Real-world analogy
A customer review such as "The phone camera is excellent, but the battery could be better" is free-form text and does not naturally fit into fixed columns.

### Python example

```python
customer_review = "The phone camera is excellent, but the battery could be better."

print("Unstructured text:")
print(customer_review)

print("Number of words:", len(customer_review.split()))
```
